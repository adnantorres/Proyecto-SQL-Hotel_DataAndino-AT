17. Hipótesis 1: ¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?

```python
%sql
SELECT
    CASE
        WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
        END AS categoria,
    avg(is_canceled) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel_dataandino.hotel_bookings
GROUP BY categoria;
```

Conclusión: Los datos confirman parcialmente la hipótesis. Las reservas realizadas por viajeros sin niños muestran una mayor propensión a la cancelación. Esto sugiere que los viajes en familia (con hijos) suelen planificarse con mayor anticipación y compromiso, lo que reduce ligeramente la tasa de cancelación en comparación con los viajeros sin cargas familiares.

18. Hipótesis 2: El impacto de las solicitudes especiales. Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes (0, 1, 2, etc.). Ordena por la cantidad de solicitudes de forma ascendente. ¿A medida que el cliente pide más cosas, cancela más o cancela menos?

```python
%sql
SELECT
    total_of_special_requests,
    avg(is_canceled) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel_dataandino.hotel_bookings
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

Conclusión: Los datos confirman que un mayor nivel de interacción, personalización o exigencia previa por parte del huésped refleja un mayor compromiso e intención real de viaje, reduciendo drásticamente el riesgo de cancelación.

19. Hipótesis 3: ¿Cambiar la reserva es señal de compromiso? Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar?

```python
%sql
SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        ELSE 'Más de 3 cambios'
        END AS agrupacion,
    avg(is_canceled) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotel_dataandino.hotel_bookings
GROUP BY agrupacion
ORDER BY tasa_cancelacion ASC;
```

Conclusión: Los datos confirman la hipótesis de que realizar modificaciones en la reserva es un indicador claro de compromiso e intención real de estadía por parte del huésped. Aquellas reservas estáticas que no experimentan ningún ajuste previo sufren un riesgo de cancelación mucho más alto.

20. Hipótesis 4: El factor precio (adr). Agrupa las reservas por tipo de hotel (hotel). Para cada tipo, muestra la cantidad total de cancelaciones y el precio promedio (adr) solo de las reservas canceladas. En base a tu consulta, ¿en qué tipo de hotel el precio promedio de cancelación es más alto?

```python
%sql
SELECT 
    hotel,
    count(*) AS total_cancelaciones,
    avg(adr) AS precio_promedio_reservas_canceladas
FROM cursodataacademy.proyecto_hotel_dataandino.hotel_bookings
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY precio_promedio_reservas_canceladas DESC;
```

Conclusión: Aunque las reservas canceladas del Resort Hotel tienen un precio promedio ligeramente superior, el City Hotel concentra el mayor volumen de cancelaciones (aproximadamente el triple). Esto sugiere que el precio promedio de las reservas canceladas no necesariamente está relacionado con la cantidad de cancelaciones.
